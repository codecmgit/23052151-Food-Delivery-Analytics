```python
from pyspark.sql import functions as F

GOLD_BASE = "/Volumes/workspace/capstone_23052151/gold_data"

restaurant_df = spark.read.format("delta").load(
    f"{GOLD_BASE}/restaurant_performance"
)

customer_df = spark.read.format("delta").load(
    f"{GOLD_BASE}/customer_segment_performance"
)

category_df = spark.read.format("delta").load(
    f"{GOLD_BASE}/category_performance"
)

print("Restaurant Performance:", restaurant_df.count())
print("Customer Segment Performance:", customer_df.count())
print("Category Performance:", category_df.count())
```

```python
# Export Gold tables as CSV

EXPORT_BASE = "/Volumes/workspace/capstone_23052151/gold_data/snowflake_export"

# Remove old export folder if it exists
dbutils.fs.rm(EXPORT_BASE, True)

# Export Restaurant Performance
restaurant_df.coalesce(1).write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(f"{EXPORT_BASE}/restaurant_performance")

# Export Customer Segment Performance
customer_df.coalesce(1).write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(f"{EXPORT_BASE}/customer_segment_performance")

# Export Category Performance
category_df.coalesce(1).write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(f"{EXPORT_BASE}/category_performance")

print("CSV export completed successfully.")
```

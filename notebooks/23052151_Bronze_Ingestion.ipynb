```python
# Unity Catalog volume paths

RAW_PATH = "/Volumes/workspace/capstone_23052151/raw_data"
BRONZE_PATH = "/Volumes/workspace/capstone_23052151/bronze_data"

print("Raw path:", RAW_PATH)
print("Bronze path:", BRONZE_PATH)
```

```python
# Check files uploaded to the raw_data volume

display(dbutils.fs.ls(RAW_PATH))
```

```python
customers_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/customers.csv")

display(customers_df)
```

```python
# ============================================================
# READ ALL RAW CSV FILES
# ============================================================

customers_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/customers.csv")

restaurants_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/restaurants.csv")

menu_items_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/menu_items.csv")

orders_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/orders.csv")

order_items_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/order_items.csv")

delivery_details_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/delivery_details.csv")

reviews_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(f"{RAW_PATH}/reviews.csv")

print("All 7 raw datasets loaded successfully.")
```

```python
# ============================================================
# RAW DATA ROW COUNTS
# ============================================================

print("Customers:", customers_df.count())
print("Restaurants:", restaurants_df.count())
print("Menu Items:", menu_items_df.count())
print("Orders:", orders_df.count())
print("Order Items:", order_items_df.count())
print("Delivery Details:", delivery_details_df.count())
print("Reviews:", reviews_df.count())
```

```python
# ============================================================
# SCHEMA CHECK
# ============================================================

print("========== CUSTOMERS ==========")
customers_df.printSchema()

print("========== RESTAURANTS ==========")
restaurants_df.printSchema()

print("========== MENU ITEMS ==========")
menu_items_df.printSchema()

print("========== ORDERS ==========")
orders_df.printSchema()

print("========== ORDER ITEMS ==========")
order_items_df.printSchema()

print("========== DELIVERY DETAILS ==========")
delivery_details_df.printSchema()

print("========== REVIEWS ==========")
reviews_df.printSchema()
```

```python
# ============================================================
# WRITE RAW DATA TO BRONZE DELTA TABLES
# ============================================================

customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/customers")

restaurants_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/restaurants")

menu_items_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/menu_items")

orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/orders")

order_items_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/order_items")

delivery_details_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/delivery_details")

reviews_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{BRONZE_PATH}/reviews")

print("All 7 datasets successfully written to Bronze.")
```
